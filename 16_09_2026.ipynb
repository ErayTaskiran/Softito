```python

class Araba:
    def __init__(self, marka, model, hiz):
        self.marka = marka
        self.model = model
        self.hiz = hiz

    def hizlandir(self, artis):
        self.hiz += artis

    def bilgi_goster(self):
        print("Marka:", self.marka)
        print("Model:", self.model)
        print("Hız:", self.hiz, "km/s")


araba1 = Araba("Ford", "Puma", 200)

araba1.bilgi_goster()

araba1.hizlandir(55)

araba1.bilgi_goster()
```

```python
class Dikdortgen:
    def __init__(self, uzunluk, genislik):
        self.uzunluk = uzunluk
        self.genislik = genislik

    def alan_hesapla(self):
        return self.uzunluk * self.genislik

    def cevre_hesapla(self):
        return 2 * (self.uzunluk + self.genislik)


dikdortgen1 = Dikdortgen(10, 5)

print("Alan:", dikdortgen1.alan_hesapla())
print("Çevre:", dikdortgen1.cevre_hesapla())
```

```python
class Kitap:
    def __init__(self, ad, yazar, sayfa_sayisi):
        self.ad = ad
        self.yazar = yazar
        self.sayfa_sayisi = sayfa_sayisi

    def ozet(self):
        return self.ad + " - " + self.yazar + " (" + str(self.sayfa_sayisi) + " sayfa)"


kitap1 = Kitap("Klon", "Eray Ünsal", 2312)

print(kitap1.ozet())
```
